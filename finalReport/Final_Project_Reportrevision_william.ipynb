**1.1 Answer:**  
Figuring out what a home is worth has always required a lot of human judgment, but that process is time-consuming and expensive. This project looked at whether machine learning could take on some of that work, using a dataset of roughly 77,600 residential properties across three California counties that Zillow originally compiled for a 2017 competition.
The dataset was far from clean. Useful features like square footage, bedroom and bathroom counts, year built, and location were all there, but so were dozens of columns that were mostly empty or completely irrelevant. Home values themselves were all over the place, with a small number of very high-value properties making the distribution difficult to work with.
Once the data was cleaned up, some clear patterns came through. Square footage was the strongest predictor of assessed value by a wide margin. Location and the age of the property also carried significant weight. Other features like lot size, garage space, and pool availability were present but added little to the model.
Three different models were trained and tuned. The best one was a Random Forest, an algorithm that builds hundreds of independent decision trees and averages their predictions to produce a more reliable result than any single tree could. It predicted assessed values within roughly $195,000 of the actual figure on average. Whether that is good enough depends on the use case. For pricing an individual home, that margin of error is hard to ignore. For a company reviewing hundreds or thousands of properties at once, it is a reasonable starting point.
This model is not a replacement for a professional appraisal. What it could do is help a real estate team work faster, surfacing properties that look mispriced and deserve a second look, while letting more straightforward cases move through the pipeline with less manual effort.

**4.2 Answer:**  
The raw dataset had 55 columns and required substantial cleaning before it was ready for modeling. Columns were dropped if they were unique property identifiers with no predictive value, had more than 70% of their data missing, or were redundant with other features already in the dataset. That left 31 features going into the modeling phase.
Missing values in the remaining columns were handled using median imputation for numerical features and most-frequent imputation for categorical ones. The median was preferred over the mean because many distributions were skewed and the mean would have been pulled by outliers. Rows where the target variable was missing were removed outright, and rows missing more than 30% of their feature values were also dropped, since imputing that much data for a single observation adds more noise than it is worth. More sophisticated imputation methods like KNN were considered but ultimately did not seem worth the added complexity, given that the models we selected are not particularly sensitive to how missing values are filled.
Outliers in the target variable were intentionally kept in. The distribution of assessed home values has a long right tail, with some properties valued in the millions, but these are legitimate observations and removing them at this stage felt like the wrong call. In hindsight this was the right decision, as the final model generalized well to the test set without any outlier removal.
Finally, categorical columns were encoded using ordinal encoding, and all features were standardized using StandardScaler fitted only on the training data to avoid leaking any information from the test set. One thing that did not work as expected was scaling itself. While it is necessary for some algorithms, it had no meaningful impact on the tree-based models we ended up using, which are not affected by feature scale.

**6.1 Answer:**  
The goal of this project was to build a model that could predict residential property assessed values from physical and location-based features. The final model, a tuned Random Forest trained on 17 cleaned features, achieved a cross-validation MAE of roughly $199,000 and a test MAE of roughly $195,000. The fact that the test MAE came in slightly lower than the cross-validation MAE is meaningful — it suggests the model generalizes well to unseen data and shows no signs of overfitting.
A few things stood out over the course of the project. Property size, location, and year built were consistently the most important predictors across every model and every feature selection method tried. Manually engineered features like log transformations and interaction terms did not improve performance, which makes sense in hindsight since tree-based models already pick up on those kinds of relationships on their own. Hyperparameter tuning helped across the board, though the gains were modest.
From a business standpoint, a model predicting within $195,000 of the actual assessed value is not precise enough to replace a professional appraisal. Where it could add real value is as a screening tool, helping a real estate team quickly identify properties that look mispriced and prioritize where human review is most needed.
The main limitations are geographic scope — the data only covers three California counties, meaning the model may not transfer reliably to other markets with different pricing dynamics. Data quality is also a constraint, since several potentially useful features were missing for a large share of properties. If this work were to continue, the most promising directions would be bringing in external data sources like neighborhood demographics or school ratings, applying a log transformation to the target variable to better handle its skewed distribution, and exploring more powerful gradient boosting libraries like XGBoost or LightGBM.
The recommendation is to deploy the Random Forest as an automated first-pass valuation tool, with human review required for high-value or unusual properties.
